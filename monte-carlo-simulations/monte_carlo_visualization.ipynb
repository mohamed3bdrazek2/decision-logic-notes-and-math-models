```python
import numpy as np
import matplotlib.pyplot as plt

def visualize_monte_carlo_pi(num_samples=10000):
    """
    Runs a Monte Carlo simulation to estimate Pi and generates a scatter plot
    visualizing the random sampling distribution.
    """
    # Generate random x and y coordinates between -1 and 1
    x = np.random.uniform(-1, 1, num_samples)
    y = np.random.uniform(-1, 1, num_samples)

    # Calculate Euclidean distance from the origin (0,0)
    # If x^2 + y^2 <= 1, the point falls inside the unit circle
    inside_circle = (x**2 + y**2) <= 1

    # The estimate is the ratio of points inside the circle to total points, multiplied by 4
    pi_estimate = 4 * np.sum(inside_circle) / num_samples

    # --- Visualization Setup ---
    plt.figure(figsize=(8, 8))

    # Plot points inside the circle in blue, outside in red
    plt.scatter(x[inside_circle], y[inside_circle], color='blue', s=1, label='Inside Circle')
    plt.scatter(x[~inside_circle], y[~inside_circle], color='red', s=1, label='Outside Circle')

    # Draw the boundary of the unit circle for clarity
    circle = plt.Circle((0, 0), 1, color='black', fill=False, linewidth=2)
    plt.gca().add_patch(circle)

    # Format the graph
    plt.title(f'Monte Carlo Pi Estimation\nSamples: {num_samples:,} | Estimated $\pi$: {pi_estimate:.5f}', fontsize=14)
    plt.xlabel('X-axis')
    plt.ylabel('Y-axis')
    plt.legend(loc='upper right')
    plt.axis('equal') # Ensures the circle doesn't look stretched like an oval
    plt.grid(True, linestyle='--', alpha=0.6)

    # Display the plot
    plt.show()

# Run the simulation with 10,000 points
visualize_monte_carlo_pi(10000)
```
